```python
# Anagram Checker
def anagram_checker(word_a, word_b):
    str1 = ''.join(sorted(word_a.lower()))
    str2 = ''.join(sorted(word_b.lower()))
    
    return str1 == str2

# Check words below:
if anagram_checker("Silent", "listen"):
    print(True)
else:
    print(False)
```

```python
anagram_checker("Silent", "Night")
```

```python
anagram_checker("night", "Thing")
```

```python
def anagram_checker(word_a, word_b, is_case_sensitive):
     if is_case_sensitive:
         str1 = ''.join(sorted(word_a))
         str2 = ''.join(sorted(word_b))
     else:
        str1 = ''.join(sorted(word_a.lower()))
        str2 = ''.join(sorted(word_b.lower()))
        
     return str1 == str2

# Check words below:
anagram_checker("Silent", "listen", False) 
```

```python
anagram_checker("Silent", "listen", True)
```

```python
anagram_checker("Silent", "Listen", True)
```

```python
anagram_checker("NIGHT", "THING", True)
```
